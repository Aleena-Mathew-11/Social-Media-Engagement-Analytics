#### Data Visualization 

##### Matplotlib

##### Scatter Plot

##### LINE PLOT : DAILY ENGAGEMENT TREND

```python
daily_trend = df.groupby(df['posted_at'].dt.date)['engagement_score'].mean()

plt.figure(figsize=(10, 5))
plt.plot(daily_trend.index, daily_trend.values, marker='o', color='green', linestyle='-')
plt.title('Daily Engagement Trend',color='b',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.xlabel('Date',color='b',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.ylabel('Average Engagement Score',color='b',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##### BAR CHART : POSTS BY CATEGORY

```python
category_counts = df['post_category'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(category_counts.index, category_counts.values, color='orange')
plt.title('Number of Posts by Category',color='purple',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.xlabel('Post Category',color='purple',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.ylabel('Count',color='purple',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.xticks(rotation=45)
plt.show()
```

##### PIE CHART : GENDER DISTRIBUTION

```python
gender_counts = df['gender'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(gender_counts.values, labels=gender_counts.index, autopct='%1.1f%%', colors=['skyblue', 'pink', 'lightgray'])
plt.title('Gender Distribution')
plt.show()
```

##### HISTOGRAM : AGE DISTRIBUTION

```python
plt.figure(figsize=(8, 5))
plt.hist(df['age'], bins=15, color='purple', edgecolor='black', alpha=0.7)
plt.title('Age Distribution of Users',color='violet',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.xlabel('Age',color='violet',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.ylabel('Frequency',color='violet',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

#### Seaborn

##### Count plot: post type 

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='post_type', palette='Set2')
plt.title('Distribution of Post Types')
plt.xlabel('Post Type')
plt.ylabel('Count')
plt.show()
```

##### BAR PLOT : AVERAGE LIKES BY CATEGORY

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x='post_category', y='likes', estimator='mean', palette='viridis', errorbar=None)
plt.title('Average Likes by Post Category')
plt.xlabel('Post Category')
plt.ylabel('Average Likes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##### VIOLIN PLOT : FOLLOWER COUNT VS SENTIMENT

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='sentiment', y='follower_count', palette='muted')
plt.title('Follower Count Distribution across Sentiments')
plt.xlabel('Sentiment')
plt.ylabel('Follower Count')
plt.show()
```

##### PAIR PLOT : NUMERIC FEATURES

```python
numeric_cols = ['likes', 'comments', 'shares', 'engagement_rate', 'sentiment']
sns.pairplot(df[numeric_cols], hue='sentiment', palette='bright')
plt.suptitle('Pair Plot of Key Numeric Features', y=1.02)
plt.show()
```

##### HEATMAP : CORRELATION MATRIX

```python
corr_matrix = df[['age', 'likes', 'comments', 'shares', 'watch_time_sec', 'impression_count', 'engagement_score']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Numeric Variables',color='purple',fontsize=20,fontweight='bold',fontstyle='italic',fontname='Times New Roman')
plt.tight_layout()
plt.show()
```
